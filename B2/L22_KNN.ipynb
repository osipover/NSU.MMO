# K ближайших соседей

K ближайших соседей (K-Nearest Neighbors, KNN) — непараметрический алгоритм, классифицирующий объект на основе классов его k ближайших соседей в пространстве признаков. Расстояние обычно вычисляется по метрике Евклида. Метод прост в реализации, не делает строгих предположений о распределении данных, но чувствителен к масштабу признаков и требует нормализации данных перед обучением.

# Используемые библиотеки

```python
import pandas as pd
import numpy as np
import kagglehub
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from sklearn.metrics import recall_score, f1_score, precision_score
```

# Описание и демонстрация датасета

Датасет представляет из себя данные пациентов. Цель модели - предсказывать пациентов, у которых диабет

```python
path_to_dataset = kagglehub.dataset_download("hassinarman/diabetes-bangladesh")
df = pd.read_csv(path_to_dataset + '/DiaBD_A Diabetes Dataset for Enhanced Risk Analysis and Research in Bangladesh.csv')

df['diabetic'] = df['diabetic'].map({'Yes': 1, 'No': 0})
y = df['diabetic']
X = df.drop(columns=['diabetic'])

print(f"Размер датасета: {df.shape}")
display(X.head())
X.info()
```

# Препроцессинг датасета

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.3)

NUM_COLUMNS = X_train.select_dtypes(include=np.number).columns.tolist()
CAT_COLUMNS = X_train.select_dtypes(include='object').columns.tolist()

# Для KNN критически важна масштабизация данных
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), CAT_COLUMNS),
        ('num', StandardScaler(), NUM_COLUMNS)  # Нормализация числовых признаков
    ]
)

# Создаём конвейер для удобства применения к train и test
pipeline = Pipeline([
    ('preprocess', preprocessor)
])

X_train_scaled = pd.DataFrame(pipeline.fit_transform(X_train))
X_test_scaled = pd.DataFrame(pipeline.transform(X_test))
X_train_scaled.columns = pipeline.named_steps['preprocess'].get_feature_names_out()
X_test_scaled.columns = pipeline.named_steps['preprocess'].get_feature_names_out()
```

# Тепловая карта

```python
plt.figure(figsize=(16, 8))
sns.heatmap(X_train_scaled.corr(), annot=True, fmt=".2f", cmap='coolwarm', annot_kws={"size": 7})
plt.title("Матрица корреляции признаков (после масштабирования)")
plt.tight_layout()
plt.show()
```

# Обучение модели

Подберем лучший k

```python
k_results = []

for k in range(3, 31):  # начинаем с 3, чтобы избежать переобучения
    model = KNeighborsClassifier(n_neighbors=k, weights='uniform')
    model.fit(X_train_scaled, y_train)

    y_pred = model.predict(X_test_scaled)
    recall = recall_score(y_test, y_pred)

    k_results.append((k, recall))

k_results = sorted(k_results, key=lambda x: x[1], reverse=True)
best_k = k_results[0][0]

print(f"Лучший k по recall: {best_k}")
```

Обучим модель с лучшим k

```python
knn_model = KNeighborsClassifier(
    n_neighbors=best_k,
    weights='uniform',
    metric='euclidean'
)

knn_model.fit(X_train_scaled, y_train)
```

Подберем лучший threshold так, чтобы recall был приемлимым, но и чтобы f1 не падала слишком сильно

```python
y_proba = knn_model.predict_proba(X_test_scaled)[:, 1]

thresholds = np.arange(0.1, 0.9, 0.01)

best_f1 = 0
best_threshold = 0

for t in thresholds:
    y_pred_t = (y_proba >= t).astype(int)

    recall = recall_score(y_test, y_pred_t)
    f1 = f1_score(y_test, y_pred_t)

    # условие: recall не слишком маленький
    if recall >= 0.6 and f1 > best_f1:
        best_f1 = f1
        best_threshold = t

print(f"Лучший threshold: {best_threshold:.2f}")
print(f"Лучший F1: {best_f1:.3f}")
```

# Тестирование модели

```python
y_pred_final = (y_proba >= best_threshold).astype(int)

print("\n=== Финальный отчёт ===")
print(classification_report(y_test, y_pred_final, target_names=['Здоров (0)', 'Диабет (1)']))
```

# Визуализация результатов

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

cm = confusion_matrix(y_test, y_pred_final)
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', ax=axes[0],
            xticklabels=['Предсказано 0', 'Предсказано 1'],
            yticklabels=['Фактически 0', 'Фактически 1'])
axes[0].set_title('Матрица ошибок (KNN)')
axes[0].set_ylabel('Фактический класс')
axes[0].set_xlabel('Предсказанный класс')

fpr, tpr, _ = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)
axes[1].plot(fpr, tpr, color='green', lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
axes[1].plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
axes[1].set_xlim([0.0, 1.0])
axes[1].set_ylim([0.0, 1.05])
axes[1].set_xlabel('False Positive Rate')
axes[1].set_ylabel('True Positive Rate')
axes[1].set_title('ROC-кривая')
axes[1].legend(loc="lower right")

plt.tight_layout()
plt.show()
```
